Colab execution uses Google's runtime.

```python
from google.colab import files
from pathlib import Path
import io
import subprocess
import sys
import tempfile
import zipfile

uploaded = files.upload()
if len(uploaded) != 1:
    raise ValueError('Upload exactly one ShipMBLang project ZIP.')
package_dir = Path(tempfile.mkdtemp(prefix='shipmblang-'))
with zipfile.ZipFile(io.BytesIO(next(iter(uploaded.values())))) as archive:
    archive.extractall(package_dir)
projects = [p.parent for p in package_dir.rglob('pyproject.toml')
            if (p.parent / 'shipmblang' / 'pipelines.py').is_file()]
if len(projects) != 1:
    raise ValueError('ZIP must contain exactly one ShipMBLang project.')
subprocess.run([sys.executable, '-m', 'pip', 'install', str(projects[0])], check=True)
```

```python
# Edit this text or paste program instructions from your notes.
source = '''Pls show me the total of 2 and 3.
Present "Keep this quotation exactly as written.".
'''
```

```python
from shipmblang import run_direct_program
import json

result = run_direct_program(source, memory=False, model_provider=None)
if result.get('status') == 'compiled' and not any(
    d.get('level') == 'error' for d in result.get('diagnostics', [])
):
    print(result.get('runtime', {}).get('stdout', ''), end='')
else:
    print(json.dumps(result, indent=2, ensure_ascii=False))
```
